# Compute metrics mean/std

```python
# mask based on the network parameters since both topologies have different masks
metrics_list = [RMSE_TEST, R2_TEST, MAE_TEST]
aux_cols = [FING_ID_COL, SUBJECT_COL, MVC_COL, SIGN_MVC_COL, PROJECT_OUT_COL,
            NORMALIZE_FORCE_COL,RNN_BIAS_COL, KERNEL_DECAY_COL, KERNEL_SIZE_COL,
            HIDDEN_STATE_SIZE_COL,
            EPOCHS_COL, RNN_TYPE_COL]
# # Compute mean and std
metrics_map_rename = {'MAE_test': 'MAE_test_mean',
                      'RMSE_test':'RMSE_test_mean',
                      'R2_test': 'R2_test_mean'}

rnn_mean_std_df = pd.DataFrame()
rnn_mean_std_df = metrics_compiled_df_rnn.groupby(aux_cols, as_index=False)[metrics_list].mean()
rnn_mean_std_df.rename(columns=metrics_map_rename, inplace=True)

# Compute std for each finger and metric (across folds)
temp_df = metrics_compiled_df_rnn.groupby(aux_cols, as_index=False)[metrics_list].std()
temp_df.rename(columns={f"{metric}": f"{metric}_std" for metric in metrics_list}, inplace=True)
rnn_mean_std_df = rnn_mean_std_df.merge(temp_df, on=aux_cols, how='left')
rnn_mean_std_df[FING_NAME_COL] = rnn_mean_std_df[FING_ID_COL].map({i: fing_order[i-1] for i in range(len(fing_order)+1)})

metrics_mean_std_list = [f"{metric}_mean" for metric in metrics_list] + [f"{metric}_std" for metric in metrics_list]
df_styled_rnn = rnn_mean_std_df.groupby([SUBJECT_COL, RNN_TYPE_COL, PROJECT_OUT_COL])[metrics_mean_std_list].mean().style.background_gradient(cmap='Blues')
df_styled_rnn_per_direction = rnn_mean_std_df.groupby([SUBJECT_COL, RNN_TYPE_COL,SIGN_MVC_COL, PROJECT_OUT_COL])[metrics_mean_std_list].mean().style.background_gradient(cmap='Blues')

df_styled_rnn_per_direction
```

# Load SNN results

```python
# load SNN data
snn_topologies = [SHALLOW_LEAKY, SHALLOW_SPIKING_DOUBLE_FILTER]
metrics_df_snn = pd.DataFrame()
for topology in snn_topologies:
    metrics_df = cr.load_and_parse_snn_data_to_df(dt=0.01, subject_list=subjects,
                                                  data_type='metrics_df_cv_snn',
                                                  network_topology=f'{topology}',
                                                  results_dir=results_dir,
                                                  config=config)
    topology_df_inf = metrics_df[metrics_df[MODE_COL] == 'inference']
    topology_df_inf[TOPOLOGY_COL] = topology
    metrics_df_snn = pd.concat([metrics_df_snn, topology_df_inf], ignore_index=True)
```

```python
metrics_df_snn.groupby([SUBJECT_COL, TOPOLOGY_COL])[SUBJECT_COL].count()
```

```python
model_short_to_full = {'intra mu spike times': 'Model 1 motor units spike times',
                       'intra mu spike times + conv + gru': 'Model 2 motor units spike times + conv + gru',
                       'intra mu spike times + conv + vanilla': 'Model 3 motor units spike times + conv + vanilla'}
```

```python
metrics_df_snn[MODEL_SHORT_NAME_COL] = metrics_df_snn.apply(lambda x:f"{x[EMG_COL]} {x[INPUT_TYPE_COL].lower()}", axis=1)
metrics_compiled_df_rnn[MODEL_SHORT_NAME_COL] = metrics_compiled_df_rnn.apply(lambda x:f"{x[EMG_COL]} {x[INPUT_TYPE_COL].lower()} + conv + {x[RNN_TYPE_COL]}", axis=1)
```

```python
metrics_compiled_df_rnn.groupby([SUBJECT_COL, RNN_TYPE_COL, MODEL_SHORT_NAME_COL])[SUBJECT_COL].count()
```

# Calculate effect size and 95% CI: comparing RNN - SNN

```python
effect_sizes_df = pd.DataFrame()
for topology in snn_topologies:
    # mask topology
    metrics_df_snn_topology = metrics_df_snn[metrics_df_snn[TOPOLOGY_COL] == topology]
    all_samples_df = pd.concat([metrics_compiled_df_rnn, metrics_df_snn_topology], axis=0, ignore_index=True)
    temp_df = cr.compute_pairwise_effect_sizes(all_samples_df, 
                                                model_short_to_full, 
                                                remove_mvcs=[],
                                                metric='RMSE_test',
                                                normalized_metric=False,
                                                seed_col=TORCH_SEED_COL)
    temp_df['topology_snn'] = topology                                            
    effect_sizes_df = pd.concat([effect_sizes_df, temp_df], axis=0, ignore_index=True)
```

```python
effect_sizes_df.groupby(['subject','model1', 'model2', 'topology_snn', 'mean_diff','mean_diff_ci_lower', 'mean_diff_ci_upper'])['mean_diff_ci_upper'].value_counts()
```
